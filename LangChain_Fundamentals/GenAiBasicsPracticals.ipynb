textSplitters
Why do we need text splitters in first place?
LLM's have token limits. Hence we need to split the text which can be large into small chunks so that each chunk size is under the token limit. There are various text splitter classes in langchain that allows us to do this.


```python
 # Say LLM token limit is 99, in that case we can do simple thing such as this
pageContent[0:99]
```

```python
# Well but we want complete words and want to do this for entire text, may be we can use Python's split funciton

words = pageContent.split(" ")

len(words)
# words 

```

```python
chunks = []

s = ""
for word in words:
    s += word + " "
    if len(s)>99:
        chunks.append(s)
        s = ""
        
chunks.append(s)
```

```python
chunks[0]
```

Splitting data into chunks can be done in native python but it is a tidious process. Also if necessary, you may need to experiment with various delimiters in an iterative manner to ensure that each chunk does not exceed the token length limit of the respective LLM.

Langchain provides a better way through text splitter classes.

Using Text Splitter Classes from Langchain
CharacterTextSplitter
RecursiveCharacterTextSplitter

```python
from langchain_text_splitters import CharacterTextSplitter

splitter = CharacterTextSplitter(
    separator ="\n",
    chunk_size = 100,
    chunk_overlap =0
)
```

```python
chunks = splitter.split_text(pageContent) #pageContent :  text
len(chunks)
```

```python
for chunk in chunks:
    print(len(chunk))
```

As you can see, all though we gave 100 as a chunk size since the split was based on \n, it ended up creating chunks that are bigger than size 200.

Another class from Langchain can be used to recursively split the text based on a list of separators. This class is RecursiveTextSplitter. Let's see how it works

```python
pageContent
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

r_splitter = RecursiveCharacterTextSplitter(
    separators = ["\n\n", "\n", " "],  # List of separators based on requirement (defaults to ["\n\n", "\n", " "])
    chunk_size = 100,  # size of each chunk created
    chunk_overlap  = 0,  # size of  overlap between chunks in order to maintain the context
    length_function = len  # Function to calculate size, currently we are using "len" which denotes length of string however you can pass any token counter)
)
```

```python
chunks = r_splitter.split_text(pageContent)
len(chunks)
chunks #printing chuncks of pageContent ... 
```

```python
for chunk in chunks : 
            print(len(chunk))
```

how RecursiveCharacterTextSplitter works ...
Recursive text splitter uses a list of separators, i.e. separators = ["\n\n", "\n", "."]

first_split = text.split("\n\n")[0]
len(first_split)

So now it will first split using \n\n and then if the resulting chunk size is greater than the chunk_size parameter which is 100 in our case, then it will use the next separator which is \n


second_split = first_split.split("\n")[0]
len(second_split)

Third split exceeds chunk size 200. Now it will further try to split that using the third separator which is ' ' (space)

second_split[2]

When you split this using space (i.e. second_split[2].split(" ")), it will separate out each word and then it will merge those chunks such that their size is close to 100{“smart packing” not random merging}

Step 1 : Create source embeddings for the text column 

```python
chunks
```

```python
from sentence_transformers import SentenceTransformer
encoder = SentenceTransformer("all-mpnet-base-v2")
vectors = encoder.encode(chunks)
print(vectors)
```

```python
vectors.shape   #(16 chunck , 768words each vector has (768dimentions))
```

```python

dim = vectors.shape[1]
dim    #size of each vector (example: 768)
```

Step 2 : Build a FAISS Index for vectors

```python
import faiss

index = faiss.IndexFlatL2(dim) #“I will store vectors of size 768 and compare them using L2Euclidean distance .”
```

Step 3 : Normalize the source vectors (as we are using L2 distance to measure similarity) and add to the index

```python
index.add(vectors) #stores your embeddings inside FAISS memory
```

```python
index   
```

 Not meant for direct viewing

```python

index.ntotal #16vectors 
```

# Step 4 : Encode search text using same encorder and normalize the output vector
                                               <!-- Similarity search (query time) -->

```python
# search_query = "I want to buy a polo t-shirt"
search_query = "What is the School of Computer Science about?"
# search_query = "What can I study at the School of Computer Science?
searchVec = encoder.encode(search_query)
searchVec.shape      #Output 768-dimensional vector  w'll not print searchVec directly cause its 768diment. Huge.. 
```

```python
# Why YOU are using it here:
# Because FAISS does NOT accept Python lists
# It wants:
#  proper numeric array format
# So we convert:
import numpy as np 
svec = np.array(searchVec).reshape(1,-1)   #convert it to numpyarr & make it 2D for FAISS
svec.shape
```

#  Retrieval: When a user poses a question, the system uses semantic search (often via vector databases like Pinecone) to find relevant, external information.


```python
distances, indices = index.search(svec, k=3) #k means top 3relevant chuncks only 
distances
indices #relevent chunck indices[]
```

```python
for i in indices[0]:
    print(chunks[i])
```
